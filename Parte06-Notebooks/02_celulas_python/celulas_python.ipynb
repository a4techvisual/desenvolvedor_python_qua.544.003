As células Python são aculativas, e reconhecem as variáveis das célula anteriores.

Observ, por exemplo, a célula abaixo:

```python
n1 = 5
print(f"O valor da variável n1 é: {n1} ")
```

O valor de n1 é claramente 5. Porém, veja na proxima celula e diga qual será o valor da variável result:

```python
result = 15 + n1
print(result)
```

O valor de result é 20. pois está somando 15 com n1, que tinha o valor de 5.
